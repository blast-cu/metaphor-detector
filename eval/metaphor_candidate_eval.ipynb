Evaluating extraction of (source verb, target noun) pairs against hand-annotations

```python
# imports
import pandas as pd
import numpy as np

from src.utils import load_env
```

```python
# set necessary variables
env_vars = load_env()
INPUT_PATH = f"{env_vars['RESULTS_DIR']}/annotation/candidate_parsing"
ANNOTATORS = ["ann0", "ann1"]
```

```python
# load annotations
annotations = []
for annotator in ANNOTATORS:
    # load annotations
    ann_filepath = f"{INPUT_PATH}/metaphor_annotations_{annotator}.csv"
    res = pd.read_csv(ann_filepath)
    annotations.append(res)
```

```python
# keep only the source verb, target noun, Sentence, and Correctly Extracted columns
ann_df = annotations[0][["Source Verb", "Target Noun", "Sentence", "Correctly Extracted"]]

# rename columns for ease of use
ann_df = ann_df.rename(
    columns={
        "Source Verb": "source_verb", "Target Noun": "target_noun", 
        "Sentence": "sentence", "Correctly Extracted": "correctly_extracted_ann0"
    }
)

ann_df["correctly_extracted_ann1"] = annotations[1]["Correctly Extracted"]
ann_df.head()
```

```python
# calculate krippendorff's alpha for inter-annotator agreement
from krippendorff import alpha

data = ann_df[["correctly_extracted_ann0", "correctly_extracted_ann1"]].replace({"Yes": 1, "No": 0}).T.values
data = data.astype(int)  # Ensure the data is of integer type
kripp_alpha = alpha(data, level_of_measurement="nominal")
print(kripp_alpha)
```

```python
# sum two columns and divide
sum = data[0] + data[1]
np.sum(sum) / 2
```

```python
bin_met_ann_df = annotations[0][["Source Verb", "Target Noun", "Sentence", "Source Metaphor"]]

# rename columns for ease of use
bin_met_ann_df = bin_met_ann_df.rename(
    columns={
        "Source Verb": "source_verb", "Target Noun": "target_noun", 
        "Sentence": "sentence", "Source Metaphor": "source_met_ann0"
    }
)

bin_met_ann_df["source_met_ann1"] = annotations[1]["Source Metaphor"]
```

```python
data = bin_met_ann_df[["source_met_ann0", "source_met_ann1"]].replace({"Yes": 1, "No": 0}).T.values
data = data.astype(int)  # Ensure the data is of integer type
kripp_alpha = alpha(data, level_of_measurement="nominal")
print(kripp_alpha)
```

```python
# sum two columns and divide
sum = data[0] + data[1]
np.sum(sum) / 2
```
