```python
import pandas as pd
import pickle
import numpy as np
import krippendorff
from sklearn.metrics import f1_score, recall_score, precision_score
from src.utils import load_env, load_json, get_logger

# variables
DATASET = "subframes_abortion"
DATASET_OUT_NAME = DATASET
env_vars = load_env()
logger = get_logger("met-class-eval")
```

```python
mc_ann_df = pd.read_csv(f"{env_vars['RESULTS_DIR']}/annotation/metaphor_classification/tweets_immigration_mc_anns.csv").drop(columns="metaphor classification_ann1").dropna()
mc_ann_df.head()
```

```python
model_res_path = load_json(f"{env_vars['RESULTS_DIR']}/metaphor_classification/llm/immigration_tweets_source_verb_target_noun_binary_met_class_qwen-teeny.json")["data"]
mc_ann_df["pred"] = mc_ann_df["id"].apply(lambda x: model_res_path[x]["annotation"]["classification"])
mc_ann_df["metaphor classification_ann0"] = mc_ann_df["metaphor classification_ann0"].astype(str)
mc_ann_df = mc_ann_df.replace({"Metaphorical": 1, "Literal": 0})
```

```python
p = round((precision_score(mc_ann_df["metaphor classification_ann0"].tolist(), mc_ann_df["pred"].tolist(), average="binary") * 100), 2)
r = round((recall_score(mc_ann_df["metaphor classification_ann0"].tolist(), mc_ann_df["pred"].tolist(), average="binary") * 100), 2)
f1 = round((f1_score(mc_ann_df["metaphor classification_ann0"].tolist(), mc_ann_df["pred"].tolist(), average="binary") * 100), 2)

f"p: {p}, r: {r}, f1: {f1}"
```
